# Template - Add New Notebook

This is a template file to add new notebooks to the tutorial.

## How to use

1. **Copy this file** with a new name following the pattern: `XX_notebook_name.ipynb`
   - Example: `04_exploratory_analysis.ipynb`, `05_cross_validation.ipynb`

2. **Edit the content** by adding code and markdown cells

3. **Add to `myst.yml`** in the `toc` section:

```yaml
toc:
  - title: "Python Practice"
    children:
      - file: notebooks/01_processamento_dados.ipynb
      - file: notebooks/02_interpolacao_krigagem.ipynb
      - file: notebooks/03_geracao_mapa_potenciometrico.ipynb
      - file: notebooks/04_new_notebook.ipynb  # ← Add here
```
