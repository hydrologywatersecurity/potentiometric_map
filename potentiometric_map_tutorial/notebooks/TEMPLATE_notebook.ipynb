# Template - Adicionar Novo Notebook

Este é um arquivo template para adicionar novos notebooks ao tutorial.

## Como usar

1. **Copiar este arquivo** com um novo nome seguindo o padrão: `XX_nome_notebook.ipynb`
   - Exemplo: `04_analise_exploratoria.ipynb`, `05_validacao_cruzada.ipynb`

2. **Editar o conteúdo** adicionando células de código e markdown

3. **Adicionar ao `myst.yml`** na seção `toc`:

```yaml
toc:
  - title: "Prática em Python"
    children:
      - file: notebooks/01_processamento_dados.ipynb
      - file: notebooks/02_interpolacao_krigagem.ipynb
      - file: notebooks/03_geracao_mapa_potenciometrico.ipynb
      - file: notebooks/04_novo_notebook.ipynb  # ← Adicionar aqui
```
